# Tensor Operations on GPU

```python
torch.cuda.is_available()  # checks if CUDA (GPU acceleration) is available on the system. It returns True if a compatible GPU is detected and can be used for computations, otherwise it returns False. This is useful for determining whether to run computations on the GPU or CPU.
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")  # creates a device object that represents the available computing device (GPU or CPU). It is useful for specifying where tensors and models should be allocated for efficient computations.
```

```python
# creating new tensors on GPU
a = torch.randn(2, 3).to(device)
a
```

```python
# moving existing tensors of cpu to GPU
b = a.to(device)
```

```python
b
```

```python
b + 5
```

```python
# code for comparing the performance of CPU and GPU for tensor operations in PyTorch. It measures the time taken to perform a large number of tensor additions on both CPU and GPU, allowing you to see the speedup achieved by using GPU acceleration for computations.
import time
import torch

#Define the size of the matrix
matrix_size = 10000 #large size to see the difference in performance between CPU and GPU

#create random matrices on CPU
matrix_cpu_1 = torch.randn(matrix_size, matrix_size)
matrix_cpu_2 = torch.randn(matrix_size, matrix_size)

# Measure time taken for matrix addition on CPU
start_time_cpu = time.time()
result_cpu = torch.matmul(matrix_cpu_1, matrix_cpu_2) #matrix multiplication on cpu
cpu_time = time.time() - start_time_cpu

print(f"Time taken for matrix multiplication on CPU: {cpu_time:.6f} seconds")

 #move matrices to GPU if available
matrix_gpu_1 = matrix_cpu_1.to(device)
matrix_gpu_2 = matrix_cpu_2.to(device)

# Measure time taken for matrix addition on GPU
start_time_gpu = time.time()
result_gpu = torch.matmul(matrix_gpu_1, matrix_gpu_2) #matrix multiplication on gpu
torch.cuda.synchronize()  # Wait for all GPU operations to finish before measuring time
gpu_time = time.time() - start_time_gpu

print(f"Time taken for matrix multiplication on GPU: {gpu_time:.6f} seconds")

#compare the results
print("\nspeedup (CPU time / GPU time):", cpu_time / gpu_time)
```
