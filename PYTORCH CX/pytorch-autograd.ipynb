```python
import torch

#inputs
x = torch.tensor(6.7, requires_grad=True)  # input feature
y = torch.tensor(0.0, requires_grad=True)   # true label(binary classification)

w = torch.tensor(1.0, requires_grad=True)  # weight
b = torch.tensor(0.0, requires_grad=True)  # bias

# Binary Cross Entropy Loss function for scalar inputs
def binary_cross_entropy_loss(prediction, target):   # actul target is zero
    epsilon = 1e-8  # small value to avoid log(0)
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)  # clamp prediction to avoid log(0)
    return -(target * torch.log(prediction + epsilon) + (1 - target) * torch.log(1 - prediction + epsilon))
```

```python
# forward bais
z = w * x + b # Weighted sum of inputs(linear parts)
y_pred = torch.sigmoid(z) # Activation function #predicted probability
```

```python
#compute binary cross entropy loss
loss = binary_cross_entropy_loss(y_pred, y)
```

```python
loss
```

```python
# back propagation

# derivatives of loss with respect to parameters
#1. dL/d(y_pred): Loss with respect tothe predicted probability(y_pred)
dL_dy_pred = (y_pred - y) / (y_pred * (1 - y_pred))

#2. d(y_pred)/dz: derivative of predicted probability with respect to z (sigmoid derivative
dY_pred_dz = y_pred * (1 - y_pred)

#3. dz/dw: derivative of z with respect to weight w and b
dz_dw = x # dz/dw=x
dz_db = 1 # dz/db=1 (bias contributes  directly to z)

dL_dw = dL_dy_pred * dY_pred_dz * dz_dw
dL_db = dL_dy_pred * dY_pred_dz * dz_db
```

```python
print(f"manual gradient for weight(dw): {dL_dw.item()}")
print(f"manual gradient for bias(db): {dL_db.item()}")
```
